<h3 align="left" style="color:#AA6C39;font-family:Algerian">TF-IDF Feature Extraction:</h3>

TF-IDF is used to convert the cleaned news headlines into numerical features that can be given to the classical machine learning models. Both unigrams and bigrams are considered to capture individual words as well as short phrases, while common English stopwords are excluded.

<a id="ngram"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 6.1 : Initialize TF-IDF Vectorizer</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
# Import TF-IDF Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize TF-IDF vectorizer
tfidf = TfidfVectorizer(
    stop_words='english',
    ngram_range=(1, 2)
)
```

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

ngram_range=(1, 2) means:
- 1 → unigrams: "apple", "government", "movie"
- 2 → bigrams: "apple iphone", "new movie", "us government"

<a id="tfidf_vec"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 6.2 : TF-IDF Vectorization</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
import joblib
X_train_tfidf = tfidf.fit_transform(X_train)

X_val_tfidf = tfidf.transform(X_val)
X_test_tfidf = tfidf.transform(X_test)

joblib.dump(tfidf, "saved_model/tfidf_vectorizer.pkl")
```

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

- fit_transform() → learns the vocabulary + calculates TF-IDF values for training data.
- transform() → uses that already-learned vocabulary and TF-IDF information for validation/test data.
- We do not fit on validation/test.

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">
After evaluating the baseline model, we tune the C hyperparameter of Logistic Regression.

C controls the strength of regularization:

- Smaller C → stronger regularization
- Larger C → weaker regularization

<a id="padding"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 10.3 : Sequence Padding</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

Headlines have different lengths, but neural networks require inputs with a consistent shape.

Therefore, we pad shorter sequences and truncate longer sequences to a fixed length.

First, let's determine a suitable sequence length from the training data:

```python
sequence_lengths = [len(seq) for seq in X_train_seq]

print("Maximum sequence length:", max(sequence_lengths))
print("Average sequence length:", np.mean(sequence_lengths))
print("95th percentile:", np.percentile(sequence_lengths, 95))
```

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

For headlines, using the 95th percentile is generally a better starting point than automatically using the absolute maximum, because a few unusually long headlines can make every sequence unnecessarily large.

```python
MAX_LEN = int(np.percentile(sequence_lengths, 95))

print("Selected sequence length:", MAX_LEN)
```

```python
X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

X_val_pad = pad_sequences(
    X_val_seq,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding='post',
    truncating='post'
)

print("Training shape:", X_train_pad.shape)
print("Validation shape:", X_val_pad.shape)
print("Test shape:", X_test_pad.shape)
```

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

Using padding='post' means zeros are added after the actual sequence.

Using truncating='post' means words beyond MAX_LEN are removed from the end.

<a id="load_glove"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 10.4 : Load GloVe Embeddings</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">

GloVe (Global Vectors for Word Representation) provides pretrained numerical representations of words based on their relationships in a large text corpus.

```python
embedding_dim = 100

glove_path = "glove.6B.100d.txt"

embeddings_index = {}

with open(glove_path, encoding='utf-8') as f:
    for line in f:
        values = line.split()
        word = values[0]
        vector = np.asarray(values[1:], dtype='float32')
        embeddings_index[word] = vector

print("Number of GloVe words:", len(embeddings_index))
```

<a id="embed_matrix"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 10.5 : Create Embedding Matrix</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
vocab_size = min(20000, len(tokenizer.word_index) + 1)

embedding_matrix = np.zeros(
    (vocab_size, embedding_dim)
)

for word, index in tokenizer.word_index.items():

    if index >= vocab_size:
        continue

    embedding_vector = embeddings_index.get(word)

    if embedding_vector is not None:
        embedding_matrix[index] = embedding_vector

print("Embedding matrix shape:", embedding_matrix.shape)
```

* The maximum headline sequence length was **43 words**, while the average was approximately **9.92 words**.
* The **95th percentile was 15 words**, so a maximum sequence length of **15** was selected to provide a compact representation while covering the majority of headlines.
* Padding produced consistent input shapes of **(62,460, 15)** for training, **(7,808, 15)** for validation, and **(7,808, 15)** for testing.
* The pretrained GloVe file was successfully loaded, containing **400,001 word vectors**.
* A **100-dimensional GloVe representation** was selected for the project.
* The resulting embedding matrix has a shape of **(20,000, 100)**, corresponding to the selected maximum vocabulary size of 20,000 and 100-dimensional word embeddings.

<h3 align="left" style="color:#AA6C39;font-family:Algerian">LSTM Model Building : </h3>
An LSTM (Long Short-Term Memory) network is a type of recurrent neural network designed to process sequential data. For this project, it processes the sequence of words in each headline and uses the information learned from the sequence to predict one of the seven news categories.

<a id="lstm_arch"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 11.1 : LSTM Architecture</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout

num_classes = len(label_encoder.classes_)

lstm_model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        weights=[embedding_matrix],
        input_length=MAX_LEN,
        trainable=False
    ),

    LSTM(128),

    Dropout(0.3),

    Dense(64, activation='relu'),

    Dropout(0.3),

    Dense(num_classes, activation='softmax')
])

lstm_model.build(input_shape=(None, MAX_LEN))

lstm_model.summary()
```

```python
lstm_model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
from tensorflow.keras.callbacks import EarlyStopping
import time

start_time = time.time()
early_stopping = EarlyStopping(
    monitor='val_accuracy',
    patience=2,
    mode='max',
    restore_best_weights=True
)

history_lstm = lstm_model.fit(
    X_train_pad,
    y_train,
    validation_data=(X_val_pad, y_val),
    epochs=10,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)
lstm_training_time = time.time() - start_time

print("LSTM Training Time:", lstm_training_time, "seconds")
lstm_model.save("saved_model/lstm.keras")
```

<h3 align="left" style="color:#AA6C39;font-family:Algerian">GRU Model Building : </h3>

The GRU model will use the same GloVe embedding matrix, padded sequences, labels, and train/validation split prepared in Steps 5 and 10. The main difference is the recurrent layer: GRU instead of LSTM.

GRU is another recurrent neural network designed to capture information from sequential data. It is similar to LSTM but has a simpler gating mechanism and generally fewer parameters.

<a id="gru_arch"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 12.1 : GRU Architecture</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, GRU, Dense, Dropout

gru_model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=embedding_dim,
        weights=[embedding_matrix],
        input_length=MAX_LEN,
        trainable=False
    ),
    GRU(128),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(num_classes, activation='softmax')
])

gru_model.build(input_shape=(None, MAX_LEN))
gru_model.summary()
```

<a id="reason_for_error"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 14.4 : Possible Reasons for Errors </span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:120%; text-align:left">


<h3 align="left" style="color:#AA6C39;font-family:Algerian">Observations :</h3>



The error analysis shows that the model made **1,118 misclassifications**, with the highest number of errors occurring in **POLITICS (245)** and **ENTERTAINMENT (230)**. The confusion matrix also shows substantial overlap between these categories, with **119 ENTERTAINMENT headlines classified as POLITICS** and **67 POLITICS headlines classified as ENTERTAINMENT**.

* **Overlapping topics:** Headlines can contain information related to more than one category. For example, entertainment headlines may discuss politicians or political events, making the distinction between ENTERTAINMENT and POLITICS difficult.

* **Shared vocabulary and entities:** Words and names associated with politics can appear in headlines from other categories. This is visible in errors such as **WOMEN → POLITICS (85)**, **QUEER VOICES → POLITICS (80)**, and **CRIME → POLITICS (73)**.

* **Short and limited context:** Headlines are often brief and may not provide enough contextual information to clearly determine their category. For example, headlines such as *“elephantback safaris”* and *“internets sad legacy no more secrets”* contain limited category-specific information.

* **Ambiguous headlines:** Some headlines can reasonably appear related to multiple topics. For example, *“meghan mccain returns to the view after father...”* was classified as ENTERTAINMENT instead of POLITICS, showing how public figures and media-related terms can overlap across categories.

* **Category-specific overlap:** **QUEER VOICES** and **WOMEN** also show notable confusion with POLITICS and ENTERTAINMENT. This suggests that social, cultural, and political topics frequently share similar vocabulary.

* **Limitations of TF-IDF features:** The Linear SVM model uses TF-IDF features, which represent the importance of individual words and word combinations but do not fully capture the deeper semantic context of a headline. Consequently, headlines with similar vocabulary may be assigned to the wrong category.

Overall, the errors are mainly associated with **overlapping topics, shared vocabulary, limited headline context, and semantic ambiguity**. These observations indicate areas where models with stronger contextual understanding could potentially improve classification performance.

```python
# Compare the improved models

final_model_comparison = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Naive Bayes',
        'Linear SVM',
        'LSTM',
        'GRU'
    ],
    'Accuracy': [
        lr_accuracy_imp,
        nb_accuracy_imp,
        svm_accuracy_imp,
        lstm_accuracy_imp,
        gru_accuracy_imp
    ],
    'Precision': [
        lr_precision_imp,
        nb_precision_imp,
        svm_precision_imp,
        lstm_precision_imp,
        gru_precision_imp
    ],
    'Recall': [
        lr_recall_imp,
        nb_recall_imp,
        svm_recall_imp,
        lstm_recall_imp,
        gru_recall_imp
    ],
    'F1-Score': [
        lr_f1_imp,
        nb_f1_imp,
        svm_f1_imp,
        lstm_f1_imp,
        gru_f1_imp
    ]
})

final_model_comparison
```

```python
best_model_name = final_model_comparison.loc[
    final_model_comparison['F1-Score'].idxmax(),
    'Model'
]

print("Selected Final Model:", best_model_name)
```

```python
# Combine training and validation data

X_train_final = pd.concat(
    [X_train, X_val],
    axis=0
)

y_train_final = pd.concat(
    [y_train, y_val],
    axis=0
)

print("Final training data shape:", X_train_final.shape)
print("Final training labels shape:", y_train_final.shape)
print("Test data shape:", X_test.shape)
print("Test labels shape:", y_test.shape)
```

```python
if best_model_name in ['Logistic Regression', 'Naive Bayes', 'Linear SVM']:

    tfidf_final = TfidfVectorizer(
        stop_words='english',
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95,
        sublinear_tf=True
    )

    X_train_final_features = tfidf_final.fit_transform(
        X_train_final
    )

    X_test_final_features = tfidf_final.transform(
        X_test
    )

    print("Selected Model:", best_model_name)
    print("Feature Type: TF-IDF")
    print("Training Shape:", X_train_final_features.shape)
    print("Test Shape:", X_test_final_features.shape)


elif best_model_name in ['LSTM', 'GRU']:

    tokenizer_final = Tokenizer(
        num_words=20000,
        oov_token="<OOV>"
    )

    tokenizer_final.fit_on_texts(X_train_final)

    X_train_final_seq = tokenizer_final.texts_to_sequences(
        X_train_final
    )

    X_test_final_seq = tokenizer_final.texts_to_sequences(
        X_test
    )

    X_train_final_features = pad_sequences(
        X_train_final_seq,
        maxlen=MAX_LEN,
        padding='post',
        truncating='post'
    )

    X_test_final_features = pad_sequences(
        X_test_final_seq,
        maxlen=MAX_LEN,
        padding='post',
        truncating='post'
    )

    print("Selected Model:", best_model_name)
    print("Feature Type: GloVe + Padded Sequences")
    print("Training Shape:", X_train_final_features.shape)
    print("Test Shape:", X_test_final_features.shape)

else:
    raise ValueError("Unknown model selected.")
```

```python
if best_model_name == 'Logistic Regression':

    final_model = LogisticRegression(
        max_iter=1000,
        random_state=42
    )

    final_model.fit(
        X_train_final_features,
        y_train_final
    )

    y_test_pred = final_model.predict(
        X_test_final_features
    )


elif best_model_name == 'Naive Bayes':

    final_model = MultinomialNB()

    final_model.fit(
        X_train_final_features,
        y_train_final
    )

    y_test_pred = final_model.predict(
        X_test_final_features
    )


elif best_model_name == 'Linear SVM':

    final_model = LinearSVC(
        random_state=42
    )

    final_model.fit(
        X_train_final_features,
        y_train_final
    )

    y_test_pred = final_model.predict(
        X_test_final_features
    )


elif best_model_name == 'LSTM':

    y_train_final_dl = y_train_final.to_numpy()

    final_model = Sequential([
        Embedding(
            input_dim=vocab_size,
            output_dim=embedding_dim,
            weights=[embedding_matrix],
            input_length=MAX_LEN,
            trainable=False
        ),
        LSTM(128, return_sequences=True),
        Dropout(0.3),
        LSTM(64),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.3),
        Dense(num_classes, activation='softmax')
    ])

    final_model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    final_model.fit(
        X_train_final_features,
        y_train_final_dl,
        epochs=10,
        batch_size=64,
        verbose=1
    )

    y_test_pred_prob = final_model.predict(
        X_test_final_features
    )

    y_test_pred = np.argmax(
        y_test_pred_prob,
        axis=1
    )


elif best_model_name == 'GRU':

    y_train_final_dl = y_train_final.to_numpy()

    final_model = Sequential([
        Embedding(
            input_dim=vocab_size,
            output_dim=embedding_dim,
            weights=[embedding_matrix],
            input_length=MAX_LEN,
            trainable=False
        ),
        GRU(128, return_sequences=True),
        Dropout(0.3),
        GRU(64),
        Dropout(0.3),
        Dense(64, activation='relu'),
        Dropout(0.3),
        Dense(num_classes, activation='softmax')
    ])

    final_model.compile(
        optimizer='adam',
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    final_model.fit(
        X_train_final_features,
        y_train_final_dl,
        epochs=10,
        batch_size=64,
        verbose=1
    )

    y_test_pred_prob = final_model.predict(
        X_test_final_features
    )

    y_test_pred = np.argmax(
        y_test_pred_prob,
        axis=1
    )

else:
    raise ValueError("Unknown model selected.")
```

<a id="prediction_func"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 17.4 : Prediction Function</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

```python
def predict_category(headline):

    if not isinstance(headline, str) or not headline.strip():
        raise ValueError("Please provide a valid headline.")

    headline = headline.lower()
    headline = re.sub(r'[^a-z\s]', '', headline)
    headline = re.sub(r'\s+', ' ', headline).strip()

    if best_model_name in [
        'Logistic Regression',
        'Naive Bayes',
        'Linear SVM'
    ]:

        headline_features = tfidf_final.transform(
            [headline]
        )

        prediction = final_model.predict(
            headline_features
        )[0]

    elif best_model_name in ['LSTM', 'GRU']:

        sequence = tokenizer_final.texts_to_sequences(
            [headline]
        )

        padded_sequence = pad_sequences(
            sequence,
            maxlen=MAX_LEN,
            padding='post',
            truncating='post'
        )

        prediction_prob = final_model.predict(
            padded_sequence,
            verbose=0
        )

        prediction = np.argmax(
            prediction_prob,
            axis=1
        )[0]

    else:
        raise ValueError("Unknown final model.")

    return label_encoder.inverse_transform(
        [prediction]
    )[0]
```

<a id="limitations"></a>
# <span style='font-family:Old English Text MT; color:#D2B48C; font-size:140%; text-align:center;'>Step 19.2 : Limitations</span>

<a href="#contents_table" style="font-family:Algerian;color:#C19A6B; font-size:100%;">Table of Contents</a>

<div style="border-radius:10px; font-family:Courier New;padding:15px; font-size:100%; text-align:left">


The classification system is limited to the seven categories selected from the HuffPost News Category Dataset. Headlines belonging to categories outside this selection may be incorrectly classified.

News headlines are often brief and may lack sufficient context to distinguish between related categories. Overlapping vocabulary, class imbalance, and variations in writing style may also affect prediction performance.

Additionally, TF-IDF features have limited contextual understanding, while LSTM and GRU models may require additional data and computational resources to improve generalization. The system classifies headlines based on learned patterns and does not verify their factual accuracy.
